# **A Revolução da Visão Computacional**

Visão Computacional usando Ultralytics YOLO! O mais aclamado modelo de deteção de objetos em tempo real.

Originalmente é um acrônimo para "You Only Live Once", que pode ser traduzido como "só se vive uma vez". Foi adaptado por Joseph Redmon para "You Only Look Once", que pode ser traduzido como "Você Só Olha Uma Vez", sendo um algoritmo de detecção de objetos extremamente eficiente e rápido em visão computacional.

Artigo de apresentação do Yolo: https://arxiv.org/abs/1506.02640

Joseph Redmon no TED: https://www.youtube.com/watch?v=Cgxsv1riJhI

O Ultralytics YOLO26 é um framework de Inteligência Artificial versátil que suporta múltiplas tarefas de visão computacional. O framework pode ser utilizado para realizar detecção, segmentação, caixas delimitadoras orientadas (OBB), classificação e estimativa de pose. Cada uma dessas tarefas possui um objetivo e um caso de uso diferentes, permitindo que você resolva diversos desafios de visão computacional com um único framework.

##**Executar detecção de objetos em Imagem**

Fazer upload de uma imagem sugerida ou uma imagem sua

```python
#!gdown "https://drive.google.com/u/0/uc?id=19y8Fjx2Ugp44lUn7kcnkrh4HZjixH2mK&export=download"
#!wget "https://raw.githubusercontent.com/scadriano/machinelearning/main/images/italia.jpg"
```

Com a imagem baixada no Google Colab Notebook, vamos usar o YOLOv26 para executar a detecção de objetos na imagem.

```python
# Execute o seguinte comando para carregar o modelo YOLOv26 pré-treinado e realizar a detecção de objetos na imagem carregada.
!yolo task=detect mode=predict model=yolo26n.pt conf=0.5 source='/content/italia.jpg'
```

O resultado da detecção de objetos será salvo no diretório "**runs/detect/predictX**".
